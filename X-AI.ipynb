```python
import os
import cv2
import numpy as np
from skimage.feature import hog
from sklearn.svm import SVC
from sklearn.metrics import classification_report, accuracy_score

# Đường dẫn đến thư mục train của dataset X-quang phổi
# (Thường cấu trúc giải nén sẽ nằm trong /content/chest_xray_dataset/chest_xray/train)
train_dir = '/content/chest_xray_dataset/chest_xray/train'

features = []
labels = []

print("Đang tiến hành trích xuất đặc trưng HOG và đọc ảnh tập Train...")

# Duyệt qua các thư mục con (NORMAL và PNEUMONIA)
categories = ['NORMAL', 'PNEUMONIA']
for label_idx, category in enumerate(categories):
    cat_path = os.path.join(train_dir, category)
    if not os.path.exists(cat_path):
        continue

    # Lấy danh sách ảnh trong thư mục (chúng ta lấy mẫu khoảng 300-500 ảnh mỗi loại trước để chạy nhanh, không bị tràn RAM)
    img_names = os.listdir(cat_path)[:300]

    for img_name in img_names:
        img_path = os.path.join(cat_path, img_name)
        # Đọc ảnh xám
        img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
        if img is not None:
            # Resize về kích thước chuẩn (ví dụ 64x64 hoặc 128x128) để tính HOG mượt mà
            img_resized = cv2.resize(img, (64, 64))

            # Trích xuất đặc trưng HOG
            hog_feat = hog(img_resized, pixels_per_cell=(8, 8), cells_per_block=(2, 2), visualize=False)

            features.append(hog_feat)
            labels.append(label_idx)

print(f"Đã trích xuất xong {len(features)} bức ảnh cho tập huấn luyện Baseline!")

# Chuyển đổi sang numpy array và huấn luyện SVM
if len(features) > 0:
    X_train = np.array(features)
    y_train = np.array(labels)

    print("Đang huấn luyện mô hình SVM (Baseline)...")
    svm_model = SVC(kernel='linear', C=1.0)
    svm_model.fit(X_train, y_train)
    print("Huấn luyện mô hình Baseline (SVM + HOG) thành công rực rỡ!")

    # Đánh giá nhanh trên tập train
    y_pred = svm_model.predict(X_train)
    print("Độ chính xác (Accuracy) trên tập mẫu Train:", accuracy_score(y_train, y_pred))
else:
    print("Không tìm thấy dữ liệu ảnh, nàng kiểm tra lại đường dẫn giúp tôi nhé!")
```

```python
import os
import cv2
import numpy as np
from skimage.feature import hog
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report

# Đường dẫn đến thư mục train bên trong dataset X-quang phổi
train_dir = '/content/chest_xray_dataset/chest_xray/train'

features = []
labels = []

print("Đang tiến hành trích xuất đặc trưng HOG cho tập Train (lấy mẫu rút gọn để chạy siêu tốc)...")

categories = ['NORMAL', 'PNEUMONIA']
for label_idx, category in enumerate(categories):
    cat_path = os.path.join(train_dir, category)
    if not os.path.exists(cat_path):
        continue

    # Lấy mẫu khoảng 300 ảnh mỗi loại để chạy mượt mà không bị tràn RAM
    img_names = os.listdir(cat_path)[:300]

    for img_name in img_names:
        img_path = os.path.join(cat_path, img_name)
        # Đọc ảnh xám
        img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
        if img is not None:
            # Resize về 64x64 để tính HOG nhanh chóng
            img_resized = cv2.resize(img, (64, 64))

            # Trích xuất đặc trưng HOG
            hog_feat = hog(img_resized, pixels_per_cell=(8, 8), cells_per_block=(2, 2), visualize=False)

            features.append(hog_feat)
            labels.append(label_idx)

print(f"Đã trích xuất thành công đặc trưng cho {len(features)} bức ảnh!")

# Huấn luyện mô hình SVM (Baseline)
if len(features) > 0:
    X_train = np.array(features)
    y_train = np.array(labels)

    print("Đang huấn luyện mô hình SVM (Baseline)...")
    svm_model = SVC(kernel='linear', C=1.0)
    svm_model.fit(X_train, y_train)
    print("🎉 Huấn luyện mô hình Baseline (SVM + HOG) thành công rực rỡ!")

    # Đánh giá trên tập huấn luyện mẫu
    y_pred = svm_model.predict(X_train)
    print("📈 Độ chính xác (Accuracy) của Baseline:", accuracy_score(y_train, y_pred))
else:
    print("⚠️ Không tìm thấy ảnh, nàng kiểm tra lại đường dẫn nhé!")
```

```python
import tensorflow as tf
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.layers import Dense, GlobalAveragePooling2D
from tensorflow.keras.models import Model
from tensorflow.keras.preprocessing.image import ImageDataGenerator

# 1. Chuẩn bị ImageDataGenerator để load ảnh trực tiếp từ thư mục train, val, test với kích thước chuẩn 224x224 cho MobileNetV2
train_dir = '/content/chest_xray_dataset/chest_xray/train'
val_dir = '/content/chest_xray_dataset/chest_xray/val'
test_dir = '/content/chest_xray_dataset/chest_xray/test'

train_datagen = ImageDataGenerator(rescale=1./255, validation_split=0.1)

train_generator = train_datagen.flow_from_directory(
    train_dir,
    target_size=(224, 224),
    batch_size=32,
    class_mode='binary'
)

# 2. Xây dựng kiến trúc MobileNetV2 (Pre-trained ImageNet)
base_model = MobileNetV2(weights='imagenet', include_top=False, input_shape=(224, 224, 3))

# Đóng băng các layers của MobileNetV2 gốc
for layer in base_model.layers:
    layer.trainable = False

# Thêm các lớp fully-connected phía trên để phân loại Nhị phân (Bình thường / Viêm phổi)
x = base_model.output
x = GlobalAveragePooling2D()(x)
x = Dense(128, activation='relu')(x)
predictions = Dense(1, activation='sigmoid')(x)

model = Model(inputs=base_model.input, outputs=predictions)

# 3. Biên dịch mô hình
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

print("Đã khởi tạo xong mô hình MobileNetV2! Sẵn sàng huấn luyện...")
```

```python
# Huấn luyện mô hình MobileNetV2 trên tập dữ liệu train
EPOCHS = 5

print("Đang bắt đầu huấn luyện mô hình MobileNetV2...")
history = model.fit(
    train_generator,
    epochs=EPOCHS,
    steps_per_epoch=len(train_generator),
    verbose=1
)

print("🎉 Huấn luyện mô hình MobileNetV2 thành công rực rỡ!")
```

```python
import os
import cv2
import numpy as np
from skimage.feature import hog
from sklearn.svm import SVC
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import classification_report, accuracy_score

print("🔍 Đang trích xuất đặc trưng HOG và chạy GridSearchCV cho SVM...")

train_dir = '/content/chest_xray_dataset/chest_xray/train'
features, labels = [], []
categories = ['NORMAL', 'PNEUMONIA']

for label_idx, category in enumerate(categories):
    cat_path = os.path.join(train_dir, category)
    if os.path.exists(cat_path):
        img_names = os.listdir(cat_path)[:300] # Lấy mẫu đại diện để chạy nhanh
        for img_name in img_names:
            img_path = os.path.join(cat_path, img_name)
            img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
            if img is not None:
                img_resized = cv2.resize(img, (64, 64))
                hog_feat = hog(img_resized, pixels_per_cell=(8, 8), cells_per_block=(2, 2), visualize=False)
                features.append(hog_feat)
                labels.append(label_idx)

X_train = np.array(features)
y_train = np.array(labels)

# Tối ưu hóa siêu tham số (Hyperparameter Tuning) bằng GridSearchCV
param_grid = {'C': [0.1, 1, 10], 'kernel': ['linear', 'rbf']}
grid_search = GridSearchCV(SVC(), param_grid, cv=3, scoring='accuracy', verbose=1)
grid_search.fit(X_train, y_train)

print(f"🎉 Siêu tham số tối ưu cho SVM: {grid_search.best_params_}")
print(f"📈 Độ chính xác tốt nhất của Baseline: {grid_search.best_score_:.4f}")
```

```python
import os
import cv2
import numpy as np
import tensorflow as tf
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.layers import Dense, GlobalAveragePooling2D
from tensorflow.keras.models import Model, load_model
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from sklearn.svm import SVC
from sklearn.model_selection import GridSearchCV
import joblib
import gradio as gr
from google.colab import drive

print("📂 Đang kết nối với Google Drive của nàng...")
drive.mount('/content/drive', force_remount=True)

# Thư mục lưu trữ trên Drive
save_dir = '/content/drive/MyDrive/Xray_Pneumonia_Project_Data'
os.makedirs(save_dir, exist_ok=True)

mobilenet_path = os.path.join(save_dir, 'mobilenetv2_final.h5')
svm_path = os.path.join(save_dir, 'best_svm_model.pkl')

# ==========================================
# 1. XỬ LÝ MÔ HÌNH MOBILENETV2 (Tự động Load từ Drive hoặc Train mới rồi lưu)
# ==========================================
if os.path.exists(mobilenet_path):
    print("⚡ Phát hiện mô hình MobileNetV2 đã lưu sẵn trên Google Drive! Đang tải về để dùng ngay...")
    model = load_model(mobilenet_path)
    print("✅ Đã tải xong MobileNetV2 từ Drive (Không cần chờ train lại)!")
else:
    print("🚀 Chưa có mô hình trên Drive. Đang tiến hành huấn luyện MobileNetV2 mới...")
    train_dir = '/content/chest_xray_dataset/chest_xray/train'

    train_datagen = ImageDataGenerator(rescale=1./255, validation_split=0.1)
    train_generator = train_datagen.flow_from_directory(
        train_dir, target_size=(224, 224), batch_size=32, class_mode='binary'
    )

    base_model = MobileNetV2(weights='imagenet', include_top=False, input_shape=(224, 224, 3))
    for layer in base_model.layers:
        layer.trainable = False

    x = base_model.output
    x = GlobalAveragePooling2D()(x)
    x = Dense(128, activation='relu')(x)
    predictions = Dense(1, activation='sigmoid')(x)

    model = Model(inputs=base_model.input, outputs=predictions)
    model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

    model.fit(train_generator, epochs=5, verbose=1)

    # Tự động lưu mô hình vừa train xong vào Drive để lần sau dùng lại
    model.save(mobilenet_path)
    print(f"💾 Đã sao lưu mô hình MobileNetV2 mới nhất sang Google Drive tại: {mobilenet_path}")

# ==========================================
# 2. XỬ LÝ MÔ HÌNH SVM + HOG TUNING (Tự động Load từ Drive hoặc Tuning mới rồi lưu)
# ==========================================
if os.path.exists(svm_path):
    print("⚡ Phát hiện mô hình Baseline (SVM) đã lưu trên Drive! Đang tải về...")
    best_svm = joblib.load(svm_path)
    print("✅ Đã tải xong mô hình SVM từ Drive!")
else:
    print("🔍 Đang chạy Tuning siêu tham số GridSearchCV cho SVM...")
    train_dir = '/content/chest_xray_dataset/chest_xray/train'
    features, labels = [], []
    categories = ['NORMAL', 'PNEUMONIA']

    if os.path.exists(train_dir):
        for label_idx, category in enumerate(categories):
            cat_path = os.path.join(train_dir, category)
            if os.path.exists(cat_path):
                img_names = os.listdir(cat_path)[:200]
                for img_name in img_names:
                    img_path = os.path.join(cat_path, img_name)
                    img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
                    if img is not None:
                        img_resized = cv2.resize(img, (64, 64))
                        from skimage.feature import hog
                        hog_feat = hog(img_resized, pixels_per_cell=(8, 8), cells_per_block=(2, 2), visualize=False)
                        features.append(hog_feat)
                        labels.append(label_idx)

        if len(features) > 0:
            X_train = np.array(features)
            y_train = np.array(labels)
            param_grid = {'C': [0.1, 1, 10], 'kernel': ['linear', 'rbf']}
            grid_search = GridSearchCV(SVC(), param_grid, cv=3, scoring='accuracy')
            grid_search.fit(X_train, y_train)
            best_svm = grid_search.best_estimator_

            # Tự động lưu SVM tốt nhất sang Drive
            joblib.dump(best_svm, svm_path)
            print(f"💾 Đã sao lưu mô hình SVM tốt nhất sang Google Drive tại: {svm_path}")

# ==========================================
# 3. GIAO DIỆN WEB TRỰC QUAN (GRADIO) ĐỂ BẬT LÊN SHOW CHO THẦY XEM
# ==========================================
def predict_xray(img):
    if img is None:
        return "⚠️ Vui lòng tải lên ảnh X-quang!"

    img_processed = tf.image.resize(img, (224, 224)) / 255.0
    img_processed = tf.expand_dims(img_processed, axis=0)

    pred = model.predict(img_processed)[0][0]

    if pred > 0.5:
        confidence = float(pred) * 100
        result = f"🔴 VIÊM PHỔI (PNEUMONIA) — Độ tự tin: {confidence:.2f}%"
    else:
        confidence = (1 - float(pred)) * 100
        result = f"🟢 BÌNH THƯỜNG (NORMAL) — Độ tự tin: {confidence:.2f}%"

    disclaimer = "\n\n⚠️ LƯU Ý Y TẾ: Sản phẩm nghiên cứu bài tập lớn, không thay thế chẩn đoán y khoa chuyên khoa."
    return result + disclaimer

demo = gr.Interface(
    fn=predict_xray,
    inputs=gr.Image(type="numpy", label="Tải lên ảnh chụp X-quang lồng ngực"),
    outputs=gr.Textbox(label="Kết quả chẩn đoán từ hệ thống AI"),
    title="🏥 HỆ THỐNG PHÂN LOẠI ẢNH X-QUANG PHỔI",
    description="Bài tập lớn Học máy — Mô hình MobileNetV2 & Giao diện Web Gradio."
)

print("\n🎉 HỆ THỐNG ĐÃ SẴN SÀNG! Đang khởi tạo giao diện Web...")
if __name__ == "__main__":
    demo.launch(share=True, inline=False)
```
